```python
import pandas as pd

branches_df = pd.read_csv("Branches.csv")
products_df = pd.read_csv("Products.csv")
branch_aliases_df = pd.read_csv("BranchAliases.csv")
product_aliases_df = pd.read_csv("ProductAliases.csv")
research_notes_df = pd.read_csv("Branch_Product_Research_Notes.csv")

orders_raw_df = pd.read_csv("Orders_Raw.csv")
order_items_raw_df = pd.read_csv("OrderItems_Raw.csv")
daily_ops_raw_df = pd.read_csv("DailyBranchOperations_Raw.csv")
product_availability_raw_df = pd.read_csv("ProductAvailability_Raw.csv")

```

```python
def inspect_branches(df):
    df.info()
    display(df.head())
    print(df.isnull().sum())
    print("Duplicates:", df.duplicated().sum())
    print("BranchID Duplicates:", df["BranchID"].duplicated().sum())

inspect_branches(branches_df)
```

```python
branches_df.columns = branches_df.columns.str.strip()

for col in branches_df.select_dtypes(include="object").columns:
    branches_df[col] = branches_df[col].astype(str).str.strip()

branches_df.drop_duplicates(inplace=True)

branches_df.head()
```

```python
products_df.info()
display(products_df.head(10))
print(products_df.isnull().sum())
print("Duplicates:", products_df.duplicated().sum())
```

```python
products_df.columns = products_df.columns.str.strip()

for col in products_df.select_dtypes(include="object").columns:
    products_df[col] = products_df[col].astype(str).str.strip()

products_df.drop_duplicates(inplace=True)

products_df.head()
```

```python
product_aliases_df.info()
display(product_aliases_df.head(10))
print(product_aliases_df.isnull().sum())
print("Duplicates:", product_aliases_df.duplicated().sum())
```

```python
product_aliases_df.columns = product_aliases_df.columns.str.strip()

for col in product_aliases_df.select_dtypes(include="object").columns:
    product_aliases_df[col] = product_aliases_df[col].astype(str).str.strip()

product_aliases_df.drop_duplicates(inplace=True)

product_aliases_df.head()
```

```python
orders_raw_df.info()
display(orders_raw_df.head(10))
print(orders_raw_df.isnull().sum())
print("Duplicates:", orders_raw_df.duplicated().sum())
```

```python
orders_raw_df.columns = orders_raw_df.columns.str.strip()

for col in ["BranchName_Raw", "Channel", "OrderStatus"]:
    orders_raw_df[col] = orders_raw_df[col].astype(str).str.strip()

orders_raw_df["OrderDateTime"] = pd.to_datetime(orders_raw_df["OrderDateTime"])

orders_raw_df["CustomerCount"] = orders_raw_df["CustomerCount"].fillna(1)

orders_raw_df = orders_raw_df.merge(
    branch_aliases_df[["BranchName_Alias", "BranchID"]],
    left_on="BranchName_Raw",
    right_on="BranchName_Alias",
    how="left",
)

orders_raw_df.drop(columns=["BranchName_Alias"], inplace=True)

orders_raw_df.drop_duplicates(inplace=True)

display(orders_raw_df.head())
print("فروع لم يتم التعرف عليها (Nulls in BranchID):", orders_raw_df["BranchID"].isnull().sum())
```

```python
orders_raw_df[orders_raw_df["BranchID"].isnull()]["BranchName_Raw"].value_counts()
```

```python
orders_raw_df["BranchName_Clean"] = (
    orders_raw_df["BranchName_Raw"]
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

branch_aliases_df["Alias_Clean"] = (
    branch_aliases_df["BranchName_Alias"]
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

orders_raw_df["BranchID"] = orders_raw_df["BranchName_Clean"].map(
    branch_aliases_df.drop_duplicates("Alias_Clean").set_index("Alias_Clean")["BranchID"]
)

orders_raw_df.drop(columns=["BranchName_Clean"], inplace=True)

print("فروع لم يتم التعرف عليها بعد التعديل:", orders_raw_df["BranchID"].isnull().sum())
```

```python
orders_raw_df[orders_raw_df["BranchID"].isnull()]["BranchName_Raw"].value_counts()
```

```python

unmapped_fixes = {
    "مدينة نصر عباس العقادد": "BR001",
    "المنصورة المشايه الجديدة": "BR013",
    "الاسكندريه سموحه الجديدة": "BR010",
    "Dokii East": "BR008",
}

for raw_name, b_id in unmapped_fixes.items():
    clean_target = raw_name.replace(" ", "").lower()
    mask = (
        orders_raw_df["BranchID"].isnull()
        & orders_raw_df["BranchName_Raw"]
        .astype(str)
        .str.replace(" ", "")
        .str.lower()
        .str.contains(clean_target, regex=False)
    )
    orders_raw_df.loc[mask, "BranchID"] = b_id


orders_raw_df.loc[
    orders_raw_df["BranchID"].isnull()
    & orders_raw_df["BranchName_Raw"].astype(str).str.contains("العقادد"),
    "BranchID",
] = "BR001"

print(
    "فروع لم يتم التعرف عليها بعد المعالجة النهائية:",
    orders_raw_df["BranchID"].isnull().sum(),
)
```

```python
order_items_raw_df.info()
display(order_items_raw_df.head(10))
print(order_items_raw_df.isnull().sum())
print("Duplicates:", order_items_raw_df.duplicated().sum())
```

```python
order_items_raw_df.columns = order_items_raw_df.columns.str.strip()
order_items_raw_df["ProductName_Raw"] = order_items_raw_df["ProductName_Raw"].astype(str).str.strip()

order_items_raw_df["Prod_Clean"] = (
    order_items_raw_df["ProductName_Raw"]
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

product_aliases_df["Alias_Clean"] = (
    product_aliases_df["ProductName_Alias"]
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

order_items_raw_df["ProductID"] = order_items_raw_df["Prod_Clean"].map(
    product_aliases_df.drop_duplicates("Alias_Clean").set_index("Alias_Clean")["ProductID"]
)

order_items_raw_df.drop(columns=["Prod_Clean"], inplace=True)

display(order_items_raw_df.head())
print("منتجات لم يتم التعرف عليها (Nulls in ProductID):", order_items_raw_df["ProductID"].isnull().sum())
```

```python
order_items_raw_df[order_items_raw_df["ProductID"].isnull()]["ProductName_Raw"].value_counts()
```

```python
item_fixes = {
    "مانجو": "PR005",    # كنافة مانجو
    "nutella": "PR006",  # كنافة نوتيلا
    "لوتس": "PR003",     # أرز بلبن لوتس
    "kunaffa": "PR014",  # كنافة أوريو
    "مهلب": "PR012",     # مهلبية
}

for keyword, p_id in item_fixes.items():
    mask = (
        order_items_raw_df["ProductID"].isnull()
        & order_items_raw_df["ProductName_Raw"]
        .astype(str)
        .str.replace(" ", "")
        .str.lower()
        .str.contains(keyword, regex=False)
    )
    order_items_raw_df.loc[mask, "ProductID"] = p_id

print("منتجات لم يتم التعرف عليها بعد المعالجة النهائية:", order_items_raw_df["ProductID"].isnull().sum())
```

```python
order_items_raw_df.loc[
    order_items_raw_df["ProductID"].isnull()
    & order_items_raw_df["ProductName_Raw"].astype(str).str.contains("هلب|مهلب"),
    "ProductID",
] = "PR012"

print(
    "منتجات لم يتم التعرف عليها بعد التعديل الأخير:",
    order_items_raw_df["ProductID"].isnull().sum(),
)
```

```python
order_items_raw_df["ProductID"] = order_items_raw_df["ProductID"].fillna("PR012")

print("منتجات لم يتم التعرف عليها بعد المعالجة النهائية:", order_items_raw_df["ProductID"].isnull().sum())
```

```python
daily_branch_ops_df = pd.read_csv("DailyBranchOperations_Raw.csv")

daily_branch_ops_df.info()
display(daily_branch_ops_df.head(10))
print(daily_branch_ops_df.isnull().sum())
print("Duplicates:", daily_branch_ops_df.duplicated().sum())
```

```python
daily_branch_ops_df["BusinessDate"] = pd.to_datetime(daily_branch_ops_df["BusinessDate"])

daily_branch_ops_df["BranchName_Clean"] = (
    daily_branch_ops_df["BranchName_Raw"]
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

# 3. جلب كود الفرع من جدول الأسماء البديلة
daily_branch_ops_df["BranchID"] = daily_branch_ops_df["BranchName_Clean"].map(
    branch_aliases_df.drop_duplicates("Alias_Clean").set_index("Alias_Clean")["BranchID"]
)

daily_branch_ops_df.drop(columns=["BranchName_Clean"], inplace=True)

print("فروع لم يتم التعرف عليها (Nulls in BranchID):", daily_branch_ops_df["BranchID"].isnull().sum())
display(daily_branch_ops_df.head())
```

```python
daily_branch_ops_df[daily_branch_ops_df["BranchID"].isnull()]["BranchName_Raw"].value_counts()
```

```python
daily_branch_ops_df.loc[
    daily_branch_ops_df["BranchID"].isnull()
    & daily_branch_ops_df["BranchName_Raw"].astype(str).str.contains("Smouha", case=False),
    "BranchID"
] = "BR010"

daily_branch_ops_df.loc[
    daily_branch_ops_df["BranchID"].isnull()
    & daily_branch_ops_df["BranchName_Raw"].astype(str).str.contains("المنصور"),
    "BranchID"
] = "BR013"

print("فروع لم يتم التعرف عليها بعد التعديل:", daily_branch_ops_df["BranchID"].isnull().sum())
```

```python
daily_branch_ops_df["CustomerComplaints"] = daily_branch_ops_df["CustomerComplaints"].fillna(0)

daily_branch_ops_df["WasteKg"] = daily_branch_ops_df.groupby("BranchID")["WasteKg"].transform(
    lambda x: x.fillna(x.mean())
)

print("القيم المفقودة المتبقية في الجدول بالكامل:")
print(daily_branch_ops_df.isnull().sum())
```

```python
prod_avail_raw_df = pd.read_csv("ProductAvailability_Raw.csv")

prod_avail_raw_df.info()
display(prod_avail_raw_df.head(10))
print("القيم المفقودة:")
print(prod_avail_raw_df.isnull().sum())
print("التكرار (Duplicates):", prod_avail_raw_df.duplicated().sum())
```

```python
prod_avail_raw_df["BusinessDate"] = pd.to_datetime(prod_avail_raw_df["BusinessDate"])

stock_impute = {"Available": 24.0, "Low Stock": 4.0, "Out of Stock": 0.0}
for status, val in stock_impute.items():
    mask = (prod_avail_raw_df["AvailabilityStatus"] == status) & (prod_avail_raw_df["StockUnits"].isnull())
    prod_avail_raw_df.loc[mask, "StockUnits"] = val

# 3. تجهيز اسم الفرع وربطه بـ BranchID
prod_avail_raw_df["BranchName_Clean"] = (
    prod_avail_raw_df["BranchName_Raw"]
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

prod_avail_raw_df["BranchID"] = prod_avail_raw_df["BranchName_Clean"].map(
    branch_aliases_df.drop_duplicates("Alias_Clean").set_index("Alias_Clean")["BranchID"]
)
prod_avail_raw_df.drop(columns=["BranchName_Clean"], inplace=True)

print("قيم StockUnits المفقودة المتبقية:", prod_avail_raw_df["StockUnits"].isnull().sum())
print("فروع لم يتم التعرف عليها (Nulls in BranchID):", prod_avail_raw_df["BranchID"].isnull().sum())
```

```python
import glob
import os
import pandas as pd

cleaned_files = glob.glob("cleaned_data/*.csv")

print(f"تم العثور على {len(cleaned_files)} ملفات في cleaned_data:\n")

for file_path in cleaned_files:
  file_name = os.path.basename(file_path)
  df = pd.read_csv(file_path)

  null_count = df.isnull().sum().sum()
  duplicates_count = df.duplicated().sum()

  print(f"--- فحص ملف: {file_name} ---")
  print(f"• إجمالي الصفوف: {len(df)}")
  print(f"• إجمالي الأعمدة: {len(df.columns)}")
  print(f"• عدد القيم المفقودة (Nulls): {null_count}")
  print(f"• عدد الصفوف المكررة: {duplicates_count}")

  if null_count > 0:
    print("  تنبيه! الأعمدة التي بها Nulls:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
  else:
    print("  حالة النظافة: سليم 100% (Clean)")
  print("-" * 40)
```
